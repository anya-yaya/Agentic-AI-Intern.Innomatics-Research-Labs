```python
#Problem 1: Smart Light Controller

class SmartLightController:
    def __init__(self, light_name, action):
        self.light_name = light_name
        self.action = action
    def display_status(self):
        print(f'\n{self.light_name} is {self.action}')

light_name = input("Enter Light Name: ")
action = input("Enter Action (ON/OFF): ")

light = SmartLightController(light_name, action)

light.display_status()
        
```

```python
#Problem 2: Employee ID Card System

class EmployeeSystem:
    def __init__(self, name, eid, dept):
        self.name = name
        self.eid = eid
        self.dept = dept
    def print_details(self):
        print("\nEmployee ID Card")
        print(f"Name: {self.name}\nID: {self.eid}\nDepartment: {self.dept}")

name = input("Enter the Employee Name: ")
eid = input("Enter the Employee ID: ")
dept = input("Enter the Employee Department: ")

employee = EmployeeSystem(name, eid, dept)
employee.print_details()
```

```python
#Problem 3: Mobile Contact Record

class ContactRecord:
    def __init__(self, name, ph_no):
        self.name = name
        self.ph_no = ph_no
    def print_details(self):
        print("\nContact Saved")
        print(f"Name: {self.name}\nPhone: {self.ph_no}")

name = input("Input Contact Name:")
ph_no = input("Input Phone Number:")

contact = ContactRecord(name, ph_no)
contact.print_details()
        
```

```python
#Problem 4: Product Price tag Generator

class RetailBillingSystem:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def print_details(self):
        print("\nBilling Details")
        print(f"Product: {self.name}\nPrice: ₹{self.price}")

prod = input("Product Name: ")
cost = input("Price: ")

product = RetailBillingSystem(prod, cost)
product.print_details()
```

```python
#Problem 5: Movie rating display system

class MovieRatingDisplaySystem:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating
    def print_details(self):
        print("\nMovie Details")
        print(f"Movie Name: {self.name}\nRating: {self.rating}/5")

name = input("Enter the movie name: ")
rating = input("Enter the ratings: ")

movie = MovieRatingDisplaySystem(name, rating)
movie.print_details()
```

```python
#Problem 6: Delivery Address Manager

class DeliveryManager:
    def __init__(self, name, addr):
        self.name = name
        self.addr = addr
    def print_details(self):
        print("\nDelivery Details")
        print("Customer Name:", self.name)
        print("Address:", self.addr)

name = input("Cutomer Name: ")
address = input("Customer Address: ")

delivery = DeliveryManager(name, address)

delivery.print_details()
```
