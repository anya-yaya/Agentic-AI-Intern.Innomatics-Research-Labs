```python
#1. User Login Check

username = input('Enter a username: ')
password = input('Enter your password: ')

if username == 'admin' and password == '1234':
    print('\nLogin Successful')
else:
    print('\nInvalid Credentials')
```

```python
#2. Pass/Fail Analyzer

marks = [45, 78, 90, 33, 60]

pass_count = 0 
fail_count = 0 

for i in marks :
    if i >= 50:
        pass_count+=1 
    else:
        fail_count+=1
        
print("Results:")
print(f' Total No. of students = {len(marks)}')
print(f' Total No. of pass students = {pass_count}')
print(f' Total No. of fail students = {fail_count}')
```

```python
#3. Simple Data Cleaner

names = [" Alice ", "bob", " CHARLIE "]

cleaned_names = [name.strip().lower() for name in names]

print("\n Cleaned Names: ", cleaned_names)
```

```python
#4. Message Length Analyzer

messages =["Hi", "Welcome to the platform", "OK"]

for i in messages:
    print(f'Length of  the message "{i}": {len(i)}')
    if len(i) > 10:
        print("⚠️ MESSAGE IS LONGER THAN 10 CHARACTERS !!")
    print()
    
```

```python
#5. Error Message Detector

logs = ["INFO", "ERROR", "WARNING", "ERROR"]

error_count = logs.count("ERROR") 
      
print("Total error count = ",error_count)
```
