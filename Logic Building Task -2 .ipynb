```python
# Problem 1: Unique words in sentence

sentence = input('Enter sentence:')

words = sentence.split()
unique_words = set(words)

print('\nUnique words count:', len(unique_words))
print('Unique words in sentence are:', unique_words)
```

```python
# Problem 2: Highest Salary from Employee data

data = {"Ravi": 75000, "Anita": 68000, "Kiran": 72000}

highest_salary = max(data.values())

for name, salary in data.items():
    if salary == highest_salary:
        employee_with_highest_salary = name
        
print(f'Highest Salary: {employee_with_highest_salary} - {highest_salary}')
```

```python
# Problem 3: Find Maximum and Minimum values

numbers = [45, 22, 89, 10, 66]

print(f'Max: {max(numbers)}')
print(f'Min: {min(numbers)}')
```

```python
# Problem 4: Count Products Above a Price Threshold

prices = [450, 1200, 899, 1500, 300]

count = 0
for price in prices:
    if price > 1000:
        count += 1
        
print(f'Products above 1000: {count}')
```

```python
# Problem 5: Calculate Attendance Percentage

attendance = ["P", "P", "A", "P", "P"]

no_of_p = attendance.count("P")
att_per = no_of_p/len(attendance)

print(f'Attendance Percentage: {att_per*100}')
```

```python
# Problem 6: Remove Duplicate Phone Numbers

phone_numbers = [9876543210, 9123456789, 9876543210]

unique_phone_nos = set(phone_numbers)

print(f'Unique Phone Numbers: {unique_phone_nos}')
```

```python
# Problem 7: Count Character Frequency

text = input("Enter a string: ")

char_freq = {}
for char in text:
    if char in char_freq:
        char_freq[char] += 1
    else:
        char_freq[char] = 1
        
print("Character Frequency:", char_freq)
```

```python
# Problem 9: Check if a Key Exists in a Dictionary

employees = {"Ravi": 75000, "Anita": 68000, "Kiran": 72000}

key = input("Enter employee name to check: ")

if key in employees:
    print(f'Employee "{key}" exists in the dictionary.')
else:
    print(f'Employee "{key}" does not exist in the dictionary.')
    
```

```python
# Problem 10: Calculate Average Marks

marks = []
n = int(input("Enter the no. of students:"))
print("\n Enter students marks:")
for i in range(n):
    marks.append(int(input()))

total = sum(marks)
avg = total/n

print("\n Average Marks: ",avg)
    
```
