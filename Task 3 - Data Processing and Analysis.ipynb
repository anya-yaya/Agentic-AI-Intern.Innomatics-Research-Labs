```python
#Problem 1: Employee Performance Bonus Eligibility

employees = {"Ravi": 92, "Anita": 88, "Kiran": 92, "Suressh": 85}

highest_score = max(employees.values())

top_performers =[]
for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)
        
print(f'Top Performers Eligible for Bonus: {", ".join(top_performers)} (Score: {highest_score})')
```

```python
#Problem 2:Search Query Analysis

import string
from collections import Counter

#Accept a string from the user, convert it to lowercase & remove punctuation
string = input("Search: ").lower().translate(str.maketrans('','', string.punctuation)) 

#Split the accepted string into word
words = string.split()

#Frequency of each word
word_count = Counter(words)

result = {}
for word, count in word_count.items():
    if count > 1:
        result[word] = count
        
print(f"Word Frequency:{result}")
```

```python
#Problem 3:Sensor Data Validation

sensor_readings =[3,4,7,8,10,12,5]

valid = []
for hr, val in enumerate(sensor_readings):
    if val%2 == 0:
        valid.append((hr, val))
        
print(f' Valid Sensor Readings (Hour, Value):{valid}')
    
```

```python
#Problem 4:Email Domain usage Analysis

emails =["ravi@gmail.com", "anita@yahoo.com", "kiran@gmail.com",
"suresh@gmail.com", "meena@yahoo.com" ]

domain_count = {}
for email in emails:
    domain = email.split("@")[1]
    domain_count[domain] = domain_count.get(domain, 0)+1

total = len(emails)
for domain in domain_count:
    percent = (domain_count[domain]/total)*100
    print(f"{domain}: {percent}%")
```

```python
#Problem 5:Sales Spike Detection

sales = [1200, 1500, 900, 2200, 1400, 3000]

avg_sales = sum(sales)/len(sales)
threshold = avg_sales * 1.3

for day, val in enumerate(sales, start=1):
    if val > threshold:
        print(f'Day {day}: {val}\n')
```
