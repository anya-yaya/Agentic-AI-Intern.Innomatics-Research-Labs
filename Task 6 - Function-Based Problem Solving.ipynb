```python
#Problem 1: Smart Parking Lot Management System

def parking_lot_management_system(capacity, logs):

    curr =0
    peak =0

    for log in logs:
        if log == "IN":
            curr +=1
        elif log == "OUT":
            curr -=1

        peak = max(peak, curr)
   
    print("Currently Parked Vehicles: ",curr)
    if curr > capacity:
        status = "Not Available"
        print("⚠️ ALERT: PARKING CAPACITY EXCEEDED !")
    elif curr == capacity:
        status = "FULL"
    else:
        status = "Available"

    print(f"Parking Status: {status}")
    
capacity = int(input("Input Total Capacity of Praking Lot: "))
logs = input("Input Vehicle Entry & Exit logs: ").split(", ")

parking_lot_management_system(capacity, logs)
```

```python
#Problem 2: Online Food Delivery Time Estimator

def food_delivery_time_estimator(d, t, w):

    speed = 40 #base speed in kmph
    time = (d/speed ) * 60 # base time in mins
    delay = 0

    if w=="rain":
        delay += 10
    elif w=="storm":
        delay += 20
    elif w=="snow":
        delay += 15
    elif w=="clear":
        delay += 0

    if t=="low":
        delay +=5
    elif t=="medium":
        delay += 10
    elif t=="high":
        delay += 20
    elif t=="extreme":
        delay +=45

    est = time + delay

    print(f"Estimated Delivery Time: {est} minutes")
    
distance = int(input("Enter the distance in km: "))
traffic = input("Enter the present traffic condition: ").lower()
weather = input("Enter the current weather condition: ").lower()

food_delivery_time_estimator(distance, traffic, weather)
```

```python
#Problem 3: Movie Theatre Seat Occupancy Analyser

def movie_booking_analyser(total, booked):
    print("No. of bookings: ",len(booked))
    per = (sum(booked)/total)*100
    print(f"Occupancy: {per:.0f}%")
    if per>=100:
        status = "FULL"
    elif 70< per <100:
        status = "ALMOST FULL"
    elif 50< per <=70:
        status = " HURRY-UP"
    elif per == 50:
        status = "HALF"
    elif per<50 :
        status = "NOT YET FULL"      

    print(f"Show Status: {status}")
    
    if booked==total:
        print("HOUSEFULL 🎉")
    if per >=90:
        print("Opening a extra show is suggested 🔔")

total = int(input("Enter Total Seats: "))
booked = list(map(int, input("Enter 1 for each bookings: ").split(",")))

movie_booking_analyser(total, booked)
    
```

```python
#Problem 4: Cloud Server Load Classification System

def server_cloud_classification(readings):
    avg_readings = sum(readings)/len(readings)
    print(f"Average CPU Load: {avg_readings:.0f}%")
    if avg_readings < 50:
        print("Server Status: Normal")
    elif 50 < avg_readings < 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")

cpu_readings = list(map(int, input("Input the CPU Readings (in % and as comma-separates values): ").split(", ")))

server_cloud_classification(cpu_readings)
```

```python
#Problem 5: Smart Classroom Resource Usage Monitor

def resource_usage_monitor(usage, set_limits):
    overused =[]
    for res in usage:
        hrs = usage[res]
        limit = set_limits.get(res, 0)
        if hrs>limit:
            overused.append(res)

    print(f"Overused Resources: {", ".join(overused)}\nEnergy Alert: {'YES' if len(overused)>0 else 'NO'}")
            
usage = eval(input("Enter the usage of classroom resource(Projector, AC & Lights) in hrs (as dictionary): "))
set_limits = {"Projector" :5, "AC" :8, "Lights" : 12}
resource_usage_monitor(usage, set_limits)
```

```python
#Problem 6: Online Event Registration Capacity Controller

def event_reg_controller(registration, capacity):
    confirmed =[]
    waitlist =[]
    for person in registration:
        if len(confirmed) < capacity:
            confirmed.append(person)
        else:
            waitlist.append(person)

    print(f"\nTotal Registrations: {len(registration)}\n")
    print(f"Confirmed Registrations: {len(confirmed)}\n{confirmed}\n")
    print(f"Waitlisted Users: {len(waitlist)}\n{waitlist}\n")
    print(f"Registration Status: {'Open' if len(confirmed)<capacity else 'Closed'}.")
    print(f"Remaining Seats: {capacity-len(confirmed)}")

capacity = int(input("Enter the Event Capacity: "))
registration = input("Enter the registrants names (comma-separated): ").split(", ")
event_reg_controller(registration, capacity)
```
