```python
#Problem 1: Mobile recharge validation System

def recharge_validation(amt, plan):
    if amt < 50:
        print("Minimum Recharge amount is Rs.50")
        return False
    if plan not in [199, 299, 3999, 499, 599]:
        print("Invalid Plan")
        return False
    if amt < plan:
        print("Recharge amount is less than selected plan.")
        return False
    return True
            
while True:
    amt = int(input("Enter your Recharge amount (must be >= Rs. 50): "))
    plan = int(input("Choose your Recharge plan (199, 299, 399, 499, 599): "))
    if recharge_validation(amt, plan):
        print("Recharge Successful ✔️\n")
        break
    else:
        print("Recharge Unsuccessful ❌. Try Again\n")
    

```

```python
#Problem 2: Inventory Reorder Alert System

stock = {"Milk": 50, "Diaper":10, "Beer": 100, "Bread": 12}

def stock_alert_system():
    for product, quantity in stock.items():
        if quantity < 15:
            print(f'{product} -> Reorder Alert 🚨 (Stock quantity:{quantity})\n')
        else:
            print(f'{product} -> Stock OK ✔️ (Stock quantity:{quantity}\n')

stock_alert_system()
    
```

```python
#Problem 3: Student Result Processing System
marks = [95, 72, 35, 100, 48]
def result_system(marks):
    total =0 
    for mark in marks:
        total += mark
    avg = total/len(marks)
    print(f'Average Marks: {avg:.2f}')
    return avg
        
result = result_system(marks)   
if result >= 50:
    print('Pass')
else:
    print('Fail')

```

```python
#Problem 4: Cab Fare Estimator with Retry Option

def cab_fare_calculator(dist, peak):
    base_fare = 50
    per_km_fare = 12

    fare = base_fare + (dist*per_km_fare)
    if peak == "y":
        fare = fare * 1.25
    return fare
    
while True:
    dist = float(input("Enter distance in km: "))
    peak = input("Is it peak hour? (y/n): ").lower()

    result = cab_fare_calculator(dist, peak)

    print(f'Total Fare: Rs. {result:.2f}')
    retry = input("Do you want to try again? (y/n): ").lower()
    if retry != "y":
        break
        
```

```python
#Problem 5: Employee Attendance Eligibility Checker

attendance = ['P', 'A', 'P', 'P', 'A', 'P']

def attendance_system(attendance):
    p_count =0 
    for day in attendance:
        if day == 'P':
            p_count += 1
    
    percentage = (p_count/len(attendance))*100

    print(f'Attendance = {percentage:.2f}%')
    
    return "Eligible" if percentage >= 75 else "Not Eligible"
       
result = attendance_system(attendance)
print(result)
```
