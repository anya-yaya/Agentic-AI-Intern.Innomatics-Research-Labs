```python
#Problem 1: Social Media- Post engagement Analyser

def analyze_social_media_engagement(likes):
    
    total = 0;

    for like in likes:
        total += like
        
    if total >=1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    print(f"Total Likes: {total}\nPost Status: {status}")

likes = list(map(int, input("Enter likes of different user, segments or platforms for the same post(comma-separated):").split(",")))

analyze_social_media_engagement(likes)
```

```python
#Problem 2: Healthcare- medicine stock alert system

def medicine_stock_alert(meds):

    for med, count in meds.items():
        if count < 10:
            status = "Low Stock Alert"
        else:
            status = "Stock Sufficient"
        
        print(f"Medicine: {med};\t Stock: {count};\t Status: {status}")

medicine_stocks = eval(input("Enter medicine stocks data as a dictionary in {}: "))

medicine_stock_alert(medicine_stocks)
    
```

```python
#Problem 3: Agriculture- Rainfall Adequency Chexker

def analyze_rainfall(data):
    n=len(data)
    req_level = 70
    avg = (sum(data)/n)
    print(avg)

    if avg >= req_level:
        print(f"Average Rainfall: {avg:.2f}\nRainfall Status: Adequate Rainfall")
    else:
        print(f"Average rainfall: {avg:.2f}\nRainfall Status: Inadequate Rainfall")

rainfall = list(map(int, input("Enter the rainfall data(comma-separated): ").split(",")))
analyze_rainfall(rainfall)
```

```python
#Problem 4: Social Media- Duplicate Account Detection

def detect_duplicate_accounts(users):

    seen = set()
    duplicates = set()

    for user in users:
        if user in seen:
            duplicates.add(user)
        else:
            seen.add(user)

    if duplicates:
        print(f"Duplicate Accounts Found: Yes\nFlagged Accounts: {", ".join(duplicates)}")
    else:
        print("Duplicated Accounts Found: No")
        
usernames = list(input("Enter usernames (comma-separated): ").split(", "))
detect_duplicate_accounts(usernames)
```

```python
#Problem 6: Agriculture- Premium Crop Price Filter

def filter_premium_crops(prices):
    threshold = 2000
    premium =[price for price in prices if price > threshold]
   
    print(f"Premium Crops: {premium}")

crop_prices = list(map(int, input("Enter the price of each crop (comma-separated): ").split(", ")))
filter_premium_crops(crop_prices)
            
```

```python
#Problem 7: System Monitoring- Application Health

def evaluate_application_health(error):
    if error == 0:
        return "Healthy"
    elif error <= 5:
        return "Minor Issues"
    elif error > 5:
        return "Critical Issues"

error_count = int(input("Enter the error count: "))
status = evaluate_application_health(error_count)
print(f"Error Count: {error_count}\nSystem Status: {status}")
```

```python
#Problem 10: Smart Electricity Bill Analyser

def calc_e_bill (units):
    # calculate bill amount
    if units<=100:
        amt = units*3
    elif 101<=units<=200:
        amt = units*5
    elif units>200:
        amt = units*7
        
    #classify usage status
    if amt<500:
        usage = "Low"
    elif 500<=amt<=1500:
        usage = "Moderate"
    elif amt>1500:
        usage = "High"

    return (amt, usage)

units = int(input("Enter the units consumed: "))
bill, classf = calc_e_bill(units)
print(f"Units consumed: {units}, Bill amount: {bill}, Status: {classf} Usage")
```
